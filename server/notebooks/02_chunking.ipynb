```python
import pandas as pd

df = pd.read_csv("../data/songs_clean.csv")

def chunk_text(text, size=150, overlap=30):
    words = text.split()
    step = size - overlap
    chunks = []
    for start in range(0, len(words), step):
        piece = words[start:start + size]
        if len(piece) < 30 and chunks:   # merge a tiny leftover into the previous chunk
            chunks[-1] += " " + " ".join(piece)
            break
        chunks.append(" ".join(piece))
    return chunks

rows = []
for song_id, row in df.iterrows():
    for i, chunk in enumerate(chunk_text(row["lyrics"])):
        rows.append({
            "song_id": song_id, "chunk_index": i, "title": row["title"],
            "artist": row["artist"], "year": row["year"], "rank": row["rank"],
            "chunk_text": chunk,
        })

chunks_df = pd.DataFrame(rows)
print(len(df), "songs ->", len(chunks_df), "chunks")
print(chunks_df["chunk_text"].str.split().str.len().describe())
print(chunks_df.iloc[0]["chunk_text"])

chunks_df.to_csv("../data/chunks.csv", index=False)
```
